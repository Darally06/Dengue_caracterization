# **Barranquilla**

Barranquilla es la capital del departamento del Atlántico, ubicada en la región Caribe de Colombia. Se sitúa estratégicamente sobre la margen occidental del río Magdalena, cerca de su desembocadura en el mar Caribe, lo que la convierte en un punto clave para el comercio marítimo y fluvial. 

Limita al norte con el municipio de Puerto Colombia y el mar Caribe, al sur con Soledad, al este con el río Magdalena y al oeste con Galapa. Esta ubicación privilegiada ha impulsado su desarrollo como centro industrial, portuario y cultural del norte colombiano.

La ciudad está dividida en cinco localidades: Riomar, Norte-Centro Histórico, Metropolitana, Sur Occidente y Sur Oriente, cada una con características urbanas y sociales distintas. Barranquilla posee un clima tropical seco, con temperaturas cálidas durante todo el año que oscilan entre los 24 °C y 33 °C. La temporada seca predomina, aunque hay lluvias entre mayo y noviembre.
